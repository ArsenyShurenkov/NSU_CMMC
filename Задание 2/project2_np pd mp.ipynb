```python
import numpy as np
import pandas as pd
def matrix(n):
    m1 = np.eye(n, k=1)
    m2 = np.eye (n, k=-1)
    m3 = 2*np.eye (n)
    return (m1 + m2 + m3)
eigenvalues, eigenvectors = np.linalg.eigh(matrix(7))

print("Собственные значения:")
for i, val in enumerate(eigenvalues):
    print(f"  λ_{i+1} = {chr(945)}+({val-2:.4f}){chr(946)}")

n = eigenvectors.shape[0]
print(f"Собственные векторы:")
for i in range(n):
    print(f"v_{i+1} = {eigenvectors[:, i]}")
```

```python
G = np.dot(eigenvectors.T, eigenvectors)
print(G)
if np.allclose(G, np.eye(n), atol = 10e-3) == True:
    print('Все хорошо')
else: print('Где-то ошибка')
```

```python
def delta_energy(n):
    eigenvalues, eigenvectors = np.linalg.eigh(matrix(n))
    λ = []
    for i, val in enumerate(eigenvalues):
        λ.append(val-2)
    p = int(round(n/2-0.5, 0))
    return(λ[p-1]-λ[p])
for i in range(3,10):
    print(f'Для n = {i}: {delta_energy(i):.4f}*{chr(946)}')
```

Она уменьшается, следовательно, на спектре поглощения с увеличением "n" будет поглощаться больший диапазон длин волн.

```python
import numpy as np
import pandas as pd
data = pd.read_csv("D:/python/ВМСХ/Задание 2/synthesis_log.csv", sep = ';')
print(f'Размер: {data.shape}')
print(f'{data.columns}')
print(data.dtypes.to_dict())
print(data['Растворитель'].value_counts(), f'\n', data['Растворитель'].nunique())
print(f'Пропусков: {data.isna().sum().to_dict()}')
```

```python
print (f'Полных дубликатов было: {int(data.duplicated().sum())}')
datan = data.drop_duplicates()
datan.reset_index(drop=True)

cols = ['Выход', 'Загрузка, ммоль', 'Чистота']

datan_clean1 = datan.copy()
for c in cols:
    datan_clean1[c] = pd.to_numeric(
        datan_clean1[c]
        .astype(str)
        .str.replace('%', '', regex=False)
        .str.replace(' ', '', regex=False)
        .str.replace(',', '.', regex=False),
        errors='coerce')

datan_clean1['Количество продукта, ммоль'] = datan_clean1['Загрузка, ммоль']*(datan_clean1['Выход']/100)*(datan_clean1['Чистота']/100)

datan_clean1['Растворитель'] = datan_clean1['Растворитель'].str.lower().str.strip()
mask_bad_loading = datan_clean1['Загрузка, ммоль'].isna()
mask_bad_res = datan_clean1['Выход'].isna()
mask_bad_pure = datan_clean1['Чистота'].isna()
broken = datan_clean1.loc[mask_bad_pure | mask_bad_res | mask_bad_loading]

datan_clean2 = datan_clean1.drop(index = broken.index).reset_index(drop=True)

datan_clean2
```

```python
y = []
x = []
for solvent in datan_clean2['Растворитель'].drop_duplicates().tolist():
    idx = datan_clean2.query("Растворитель == @solvent").index
    values = datan_clean2.loc[idx, 'Выход'].to_numpy()
    y.append(solvent)
    x.append(np.mean(values))
x = np.array(x)
df = pd.DataFrame()
df['Растворитель'] = y
df['Средний выход'] = x
print(df)
print(f'Растворитель с наименьшим средним выходом: {y[np.argmax(x == x.min())]}, средний выход = {x.min():.3f}')
print(f'Растворитель с наибольшим средним выходом: {y[np.argmax(x == x.max())]}, средний выход = {x.max():.3f}')
```

```python
import matplotlib.pyplot as plt
from matplotlib.ticker import MultipleLocator
from scipy.optimize import curve_fit
fig, ax = plt.subplots()
ax.bar (y, x, color = "r")
ax.set_title('Средний выход растворителей')
ax.set_ylabel('Средний выход, %')
plt.show()
```
