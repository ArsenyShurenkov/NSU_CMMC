Чистка данных/

```python
measurements = [10.1, 10.2, 10.0, 9.9, 10.1, 10.0, 10.2, 9.8, 10.1, 10.0, 10.1, 10.2, 9.9, 10.0, 25.0]
avg = round(1/len(measurements)*sum(measurements), 2)
delta = []
for i in measurements:
    delta.append((i-avg)**2)
s = round((1/(len(measurements)-1)*sum(delta))**0.5, 2)
print(f"--Исходные данные --\nСреднее: {avg}\nСтд. отклонение (s): {s}")
delta = []
delete = []
for i in measurements:
    x = (i-avg)**2
    if x**0.5 > 3*s:
        delete.append(i)
        measurements.remove(i)
avg = round(1/len(measurements)*sum(measurements), 2)
for i in measurements:
    delta.append((i-avg)**2)
avg = round(1/len(measurements)*sum(measurements), 2)
s = round((1/(len(measurements)-1)*sum(delta))**0.5, 2)
print (f"Обнаружено выбрасов: {len(delete)} (Значения): {delete}")
print(f"--Новые данные --\nСреднее: {avg}\nСтд. отклонение (s): {s}")
```

```python
sets1 = {"Temp": ["295", "300", "305"], "Sample": ["Al2O3", "SiO2", "MgO", "NaF"], "Success": [True, False]}
sets2 = {"Concentration": [1e-5, 2e-5, 3e-5, 4e-5, 5e-5], "Sample": ["CuSO4", "KMnO4", "Ni(NO3)2"], "Exp_type":["Baseline", "Exp"], "Device": ["Cary4000", "SF56"]}
def f(data):
    if not data:
        return []
    result = None
    for key in data:
        if result is None:
            result = [[str(d)] for d in data[key]]
            continue
        new_result = []
        for r in result:
            for d in data[key]:
                new_result.append(r + [str(d)])
        result = new_result
    return [" ".join(combo) for combo in result]
print(f(sets1))
print(f(sets2))
```
